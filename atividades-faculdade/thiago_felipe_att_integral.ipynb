`THIAGO FELIPE RIBEIRO BRITO` INTEGRAIS POR METODO DE **SUBISTITUIÇÃO**

```python
import sympy as sp
import matplotlib.pyplot as plt
import numpy as np

# Variável simbólica
x = sp.symbols('x')

# Lista de integrais (função, intervalo, resposta esperada)
integrals = [
    (2*x*(x**2 + 1)**3, (0, 1), 15/4),
    (sp.tan(x)*sp.sec(x)**2, (0, sp.pi/4), 1/2),
    (sp.log(x)/x, (1, sp.E), 1/2),
    (sp.sin(3*x)*sp.cos(x), (0, sp.pi/2), 1/4),
    (x*sp.exp(x**2), (0, 1), (sp.E - 1)/2),
    (1/sp.sqrt(1-x**2), (0, 1), sp.pi/2),
    (x**3*sp.sqrt(1+x**4), (-1, 1), 0),
    (sp.sin(2*x), (0, sp.pi/2), 1),
    (((1-x**2)**2)/(1+x**2), (0, 1), sp.pi/4 - 1/2),
    (sp.exp(sp.log(x)**2)/x, (1, sp.E), sp.E - 1),
    (3*x**2*sp.exp(x**3), (0, 1), sp.E - 1),
    (sp.sec(x)**2, (0, sp.pi/4), 1)
]

def calcular_e_plotar(integrals):
    for i, (f, intervalo, esperado) in enumerate(integrals, start=1):
        # Calcular integral
        integral = sp.integrate(f, (x, intervalo[0], intervalo[1]))
        integral_num = integral.evalf()

        # Mostrar resultado
        print(f"Integral {i}:")
        print(f"Função: {f}")
        print(f"Intervalo: {intervalo}")
        print(f"Resultado calculado: {integral_num}")
        print(f"Resultado esperado: {esperado}\n")

        # Preparar dados para plot
        xs = np.linspace(float(intervalo[0]) + 1e-4, float(intervalo[1]) - 1e-4, 300)
        f_lambd = sp.lambdify(x, f, "numpy")

        try:
            ys = f_lambd(xs)
        except Exception as e:
            print(f"⚠ Erro ao avaliar função da integral {i}: {e}")
            continue

        # Plotar
        plt.figure(figsize=(5,3))
        plt.plot(xs, ys, label=f'Integral {i}', color = 'darkred')
        plt.fill_between(xs, ys, alpha=0.2)
        plt.title(f"Integral {i}")
        plt.xlabel("x")
        plt.ylabel("f(x)")
        plt.legend()
        plt.grid(True)
        plt.show()

# Executar
calcular_e_plotar(integrals)
```
