```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

# Variável simbólica
x = sp.symbols('x')

# Funções e intervalos
integrals = [
    (3*x**2, 0, 2), (2*x + 1, 0, 4), (x**3, 0, 1), (sp.sin(x), 0, sp.pi),
    (sp.cos(x), 0, sp.pi/2), (1/x, 1, sp.E), (sp.exp(2*x), 0, 1), (sp.tan(x), 0, sp.pi/4),
    (1/(1+x**2), 0, 1), (sp.sqrt(x), 0, 1), (x**5, -1, 1), (x**4, -2, 2),
    (sp.cos(2*x), 0, sp.pi), (2*x/(x**2 + 1), 0, sp.ln(2)), (sp.exp(x), 0, 1),
    (3*x**2 + 4, 0, 1), (1/x**2, 1, 2), (sp.sin(x)*sp.cos(x), 0, sp.pi/6),
    ((x+1)/(x+1)**2, 0, 1), (4*x+1, 0, 2)
]

# Calcular e plotar
for i, (f, a, b) in enumerate(integrals, 1):
    # Integral
    result = sp.integrate(f, (x, a, b))
    print(f"Integral {i}: {result}")

    # Plot
    X = np.linspace(float(a), float(b), 200)
    Y = sp.lambdify(x, f, 'numpy')(X)

    plt.figure(figsize=(5,3))
    plt.plot(X, Y, label=f'f(x) = {f}')
    plt.fill_between(X, Y, alpha=0.2)
    plt.title(f'Integral {i}: ∫f(x)dx de {a} a {b}')
    plt.xlabel('x')
    plt.ylabel('f(x)')
    plt.grid(True)
    plt.legend()
    plt.tight_layout()
    plt.show()
```
