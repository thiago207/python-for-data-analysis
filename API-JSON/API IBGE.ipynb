### API IBGE

https://servicodados.ibge.gov.br/api/docs

```python
import requests
import pandas as pd
import matplotlib.pyplot as plt


link = 'https://servicodados.ibge.gov.br/api/v3/agregados/4112/periodos/2006/variaveis/2586?localidades=N1[all]'

r = requests.get(link)
infos = r.json()

display(infos)
```
