#### Consulta de informações de um CEP

```python
import requests
import json
import pandas

cep = '54803073'

link = f'https://viacep.com.br/ws/{cep}/json/'

requisicao = requests.get(link)
requisicao_dir = requisicao.json()

display(requisicao_dir)
```

#### Busca de CEP a partir de endereço

```python
import pprint
uf = 'PE'
cidade = 'Moreno'
endereco = 'Rua Joaquim Nabuco'

link = f'https://viacep.com.br/ws/{uf}/{cidade}/{endereco}/json/'

requisicao = requests.get(link)
requisicao_dir = requisicao.json()

for i in requisicao_dir:
    print(f'{i['cep']}')
    
```
