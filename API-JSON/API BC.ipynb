### API do Banco Central

https://dadosabertos.bcb.gov.br/dataset?res_format=API

```python
import requests
import json
import pandas

link = "https://olinda.bcb.gov.br/olinda/servico/mecir_dinheiro_em_circulacao/versao/v1/odata/informacoes_diarias?$top=10000&$orderby=Data%20desc&$format=json"

requisicao = requests.get(link)
informacoes = requisicao.json()

display(informacoes)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame(informacoes['value'])

#tabela["Valor"] = tabela["Valor"].map("R${:,.2f}".format)

agrupado_por_especie = df.groupby('Especie')['Valor'].sum().reset_index()

display(agrupado_por_especie)

plt.bar(df['Especie'], df['Valor'], color=['#3b82f6', '#10b981'], width=0.5)

# Configurações
plt.ylabel('Valor (R$)', fontsize=12, fontweight='bold')
plt.title('Comparação: Cédulas vs Moedas', fontsize=14, fontweight='bold')
plt.grid(True, alpha=0.3, axis='y')

plt.tight_layout()
```
